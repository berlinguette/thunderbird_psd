# TOML Example
This notebook shows an example of how `.TOML` files can be read and used using the `tomli` library. 

When opening file ensure that the `.TOML` file is read as binary with the flag `"rb"` as per documentation. Then, all entries in the file can be accessed easily as dictionary fields!

**Note**: Comments (text marked with preceding '#') in the `.toml` file will not be parsed.

```python
import tomli
from pathlib import Path
```

```python
TOML_PATH = Path("../sample_datasets/20220824_CERC_background/exp_info.toml")
```

```python
# Must open with "rb" as stated in documentation
with open(TOML_PATH, "rb") as f:
    exp_info = tomli.load(f)
```

```python
exp_info["exp_id"]
```
