```python
from pathlib import Path
import pandas as pd
from data_processing.arc_paths import get_report_root
import matplotlib.pyplot as plt
```

```python
exp_name = "ID-479"
file_name = f"{exp_name}_data_300s_bin.csv"
exp_output_path = get_report_root(exp_name)
exp_data_path = exp_output_path / file_name
```

```python
df = pd.read_csv(exp_data_path)
df
```

```python
df['Total rate (cps)'] = df['Neutron rate (cps)'] + df['Background gamma rate (cps)']
df
```

```python
total_rate_series = df['Total rate (cps)']
```

```python
rate_mean = total_rate_series.mean()
rate_mean
```

```python
rate_std_dev = total_rate_series.std()
rate_std_dev
```

```python
# TODO get range of total rate
```

```python
print(f"Mean rate: {rate_mean:.2f} cps")
print(f"Std. dev.: {rate_std_dev:.2f} cps")
print(f"Relative:  {rate_std_dev / rate_mean:.2%}")
```
