```python
import pandas as pd
import numpy as np
from scipy.optimize import root_scalar
import matplotlib.pyplot as plt
```

```python
def cal_curve(x):
    a = -3.087
    b = 3.146e+7
    c = 710.7
    return a*np.log(b*x+1)+c*x
```

```python
l_data = np.linspace(0, 30, 500)
ph_data = cal_curve(l_data)
fig, ax = plt.subplots(figsize=(8, 8))
ax.plot(
    ph_data,
    l_data,
)
ax.set_xlim(100, 16500)
ax.set_ylim(None, 25)
ax.set_xlabel("Pulse Height (ADC)", fontsize=14)
ax.set_ylabel("Light energy (MeVee)", fontsize=14)
```

```python
cal_curve(30)
```

```python
bracket = (0, 30)


def invert_cal_curve(ph):
    sol = root_scalar(
        lambda ee: cal_curve(ee) - ph,
        bracket=bracket,
        method="brentq"
    )
    return sol.root


invert_cal_curve_vec = np.vectorize(invert_cal_curve)

ph_data_inv = np.linspace(100, 16500, 500)
l_data_inv = invert_cal_curve_vec(ph_data_inv)
print(l_data_inv[:5])

fig, ax = plt.subplots(figsize=(8, 8))
ax.plot(
    ph_data_inv,
    l_data_inv,
)
ax.set_xlim(100, 16500)
ax.set_ylim(None, 25)
ax.set_xlabel("Pulse Height (ADC)", fontsize=14)
ax.set_ylabel("Light energy (MeVee)", fontsize=14)
```

```python
ph_data_round_trip = cal_curve(l_data_inv)
delta = ph_data_round_trip - ph_data_inv
print(delta[:5])

fig, ax = plt.subplots(figsize=(8, 8))
ax.plot(
    ph_data_inv,
    delta,
)
ax.set_xlim(100, 16500)
# ax.set_ylim(None, 25)
ax.set_yscale("log")
ax.set_xlabel("Pulse Height (ADC)", fontsize=14)
ax.set_ylabel("Delta with cal_curve(inv_cal_curve(PH))", fontsize=14)
```
