This notebook is designed to merge all Parquet files in a converted experiment folder.

## Initialization

```python
# Imports
import pandas as pd

from data_processing import loading as load
from data_processing import processing as proc
from data_processing import helpers as help
from data_processing import arc_paths
```

```python
def get_exp_processed_data_path(exp_id):
    exp_converted_path = arc_paths.get_exp_root(exp_id)
    return exp_converted_path / "processed_data" / "unfiltered"


def get_psd_folder_path(exp_id):
    exp_processed_path = get_exp_processed_data_path(exp_id)
    return exp_processed_path / "psd"


def get_signals_folder_path(exp_id):
    exp_processed_path = get_exp_processed_data_path(exp_id)
    return exp_processed_path / "signals"
```

## Experiment ID Input

```python
experiment_ids = help.input_experiment_ids()
```

```python
# Find output path
all_parquet_data = {}
for exp_id in experiment_ids:
    exp_psd_folder_path = get_psd_folder_path(exp_id)
    exp_signals_folder_path = get_signals_folder_path(exp_id)
    psd_df = pd.read_parquet(exp_psd_folder_path)
    signals_df = pd.read_parquet(exp_signals_folder_path)
    exp_parquet_data = {"psd": psd_df, "signals": signals_df}
    all_parquet_data[exp_id] = exp_parquet_data
```

```python
for exp_id, exp_parquet_data in all_parquet_data.items():
    psd_df = exp_parquet_data['psd']
    signals_df = exp_parquet_data['signals']
    # print(psd_df)
    # print(signals_df)
    # print(psd_df.index.equals(signals_df.index))
    psd_merge_path = get_psd_folder_path(exp_id) / "merged"
    psd_merge_path.mkdir(parents=True, exist_ok=True)
    signals_merge_path = get_signals_folder_path(exp_id) / "merged"
    signals_merge_path.mkdir(parents=True, exist_ok=True)
    psd_df.to_parquet(psd_merge_path / f"{exp_id}_merged_psd.parquet", index=True)
    signals_df.to_parquet(signals_merge_path / f"{exp_id}_merged_signals.parquet", index=True)
```
